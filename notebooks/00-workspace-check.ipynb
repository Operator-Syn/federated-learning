# Workspace check

Smoke test for the `fl-thesis` environment. Run all cells once after `uv sync`; if it
executes clean, the kernel, the package import and the plotting path are wired correctly.

Cells here hold **configuration and calls only**. The logic lives in `src/fl_thesis`,
split by what each part is responsible for:

| subpackage | module | responsibility |
|---|---|---|
| `workspace` | `paths` | locate the repository, create `runs/` |
| `runtime` | `environment` | interpreter, host and library versions |
| `data` | `synthetic` | reproducible placeholder series |
| `viz` | `figures` | build and save figures (Agg) |

That split exists because notebook outputs are ephemeral — the Git clean filter strips
them before commit — so anything worth keeping has to be a file written by code, and
anything worth reusing has to be importable.


```python
# Config: imports and the values this notebook sets. No computation here.
from IPython.display import Image, display

from fl_thesis.data.synthetic import availability_series
from fl_thesis.runtime.environment import collect
from fl_thesis.viz.figures import availability_figure, save_figure
from fl_thesis.workspace.paths import repo_root, runs_dir

ROUNDS = 20
SEED = 20261118
ROOT = repo_root()
OUTPUT_DIR = runs_dir("00-workspace-check")
```

```python
# Call: where am I actually running?
report = collect()
print("\n".join(report.format_lines()))
```

```python
# Call: build from the placeholder series and write the artifact to runs/
stations = availability_series(ROUNDS, seed=SEED)
figure_path = save_figure(
    availability_figure(stations, title=f"synthetic: availability over {ROUNDS} rounds"),
    OUTPUT_DIR / "availability.png",
)
print("wrote", figure_path.relative_to(ROOT))

# show the file that was written; a stripped cell output never hides the result
display(Image(filename=str(figure_path)))
```
